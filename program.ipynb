```python
# Importing Libraries
import pandas as  pd 
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
# Load the dataset
iris = load_iris()
```

```python
# Create dataframe from iris dataset
df=pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species']= iris.target
```

```python
df.head()
```

```python
# Split the dataset intp features (X) and target (y)
X= df.drop('species', axis=1)
y= df['species']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3,random_state=42)
```

```python
# train Decision tree Classifier
classifier = DecisionTreeClassifier()
classifier.fit(X_train, y_train)
```

```python
# make prediction
y_pred = classifier.predict(X_test)
y_pred
```

```python
# Evalute Model using accuracy, confusion matrix and classification report
acc =metrics.accuracy_score(y_test,y_pred)
conf_matrix = metrics.confusion_matrix(y_test, y_pred)
class_report = metrics.classification_report(y_test,y_pred)
```

```python
print(f"accuracy{acc}")
print(f"confusion matrix {conf_matrix}")
print(f"class report{class_report}")
```

```python
# Visualize the Decisin tree
plt.figure(figsize=(12,8))
plot_tree(classifier, feature_names=iris.feature_names,class_names=iris.target_names, filled=True)
plt.title("Desion tree for Iris flower Classification")
plt.show()
```
